- create a class book with title, author, and price attributes.
- Demonstrate encapsulation by protecting the price attribute.
- Use inheritace to create an EBook subclass.

```python
# create a class book with title, author, and price attributes.
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self._price = price    # protected attribute (encapsulation)
        self.set_price
    def get_price(self):        
        return self._price

    def set_price(self, price):   
        if price < 0:
            raise ValueError("Price cannot be negative.")
        self._price = price

    def get_info(self):
        return f"'{self.title}' by {self.author} - ${self._price:.2f}"
```

```python
#  Use inheritace to create an EBook subclass.
class Ebook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size  # in MB

    def get_info(self):
        return super().get_info() + f" [EBook, {self.file_size} MB]"
```

```python
book = Book("Python Principles", "JIN", 35.5)
ebook = Ebook("How to Write Clean Code", "YURI", 20, 4.2)

print(book.get_info())
print(ebook.get_info())
print("Price (via getter):", book.get_price())
```
